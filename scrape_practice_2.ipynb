```python
# import requests
import requests

# Function to fetch the webpage. This function fetches the HTML content of the specified URL and handles any request errors
def fetch_webpage(url):
    try:
        response = requests.get(url)
        response.raise_for_status()    # Raise an error for bad responses
        return response.text
    except requests.RequestException as e:
        print(f"Error fetching the webpage: {e}")
        return None
```

```python
# Import BeautifulSoup
from bs4 import BeautifulSoup
def parse_webpage(html, base_url):
    soup = BeautifulSoup(html, 'html.parser')
    articles = []
    
    # Find all the "card" elements representing the articles
    for card in soup.select('.card.container__item'):
        headline_tag = card.select_one('.container__headline-text')  # Headline text
        link_tag = card.select_one('.container__link')  # Article link
        
        if headline_tag and link_tag:
            headline = headline_tag.get_text(strip=True)  # Extract the headline text
            article_url = link_tag['href']  # Extract the URL from the 'href' attribute
            
            # If URL is relative, concatenate with the base URL
            if article_url.startswith('/'):
                article_url = base_url + article_url
            
            # Append the data to the articles list
            articles.append({'headline': headline, 'url': article_url})
    return articles
```

```python
# Combining everything in the main function
def main():
    url = 'https://edition.cnn.com/'    # URL of the target scrape website
    html_content = fetch_webpage(url)     # Fetch the webpage
    if html_content:    # If fetch was successful
        articles = parse_webpage(html_content, base_url)   # Parse the quotes 
        for article in articles:
            print(f"Headline: {article['headline']}\nURL: {article['url']}\n")
if __name__ =='__main__':
    base_url = 'https://edition.cnn.com'
    starting_url = 'https://edition.cnn.com/?refresh=1'
    main()    # Execute the main function
```
